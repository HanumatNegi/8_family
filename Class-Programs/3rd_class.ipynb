```python
x = 1
while x <= 100:
    if x % 3 == 0 and x % 5 == 0:
        print("FizzBuzz")
    elif x % 3 == 0:
        print("Fizz")
    elif x % 5 == 0:
        print("Buzz")
    else:
        print(x)
    x += 1


#FizzBuzz #Fizz #Buzz #combination of fizz buzz or integers
```

```python
#Calculate sum of all numbers from 1 to N using a for loop.

num = int(input('Enter a number : '))
sum = 0

for i in range(1,num+1):
  sum = sum + i

print(sum)
```
